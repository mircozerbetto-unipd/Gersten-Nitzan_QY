# **1 Introduction**
This notebook implements the calculation of the quantum yeld of emitting molecules in proximity to spherical metal nanoparticles (MNP). The parallel (radial), and perpendicular (tangential) components are computed, as well as their average over the whole solid angle. Basic equations are found in Gersten, J.; Nitzan, A. J. Chem. Phys. 1981, 75, 1139-1152 ([DOI:10.1063/1.442161](https://doi.org/10.1063/1.442161)). The simplified expressions for a sphere have been implemented.

## **1.A Geometry**
The metal nanoparticle has a diameter $2a$, with $a$ the radius. The geometric center of the metal NP is on the origin of the reference frame. The emitter is positioned at a certain distance $d$ from the surface of the sphere, i.e. it is ad a distance $r=a+d$ from the origin of the reference system.

The molecular electric dipole of the emitter is considered with two orientations: parallel (tangential) or perpendicular (radial) to the surface of the nanoparticle.

## **1.B Optical properties**
The emitter electrical polarizability is corrected including radiative damping and dynamic depolarization:

$\alpha=\displaystyle{
  \frac{\alpha_0}{1-\frac{k^2 a^2}{10}-i\frac{k^3\alpha_0}{6\pi}}
}\,\,\,\,\,\,\,\,\,\,(1)$

with

$\alpha_0 = a^3 \displaystyle \frac{\epsilon_\textrm{p}'-1}{\epsilon_\textrm{p}'+2}\,\,\,\,\,\,\,\,\,\,(2)$

the dipole polarizability of the MNP, $\epsilon_\textrm{p}'=\epsilon_\textrm{p}/\epsilon_\textrm{m}$, $\epsilon_p(\omega)$ is the MNP frequency dependent dielectric function, $\epsilon_m$ the dielectric function of the medium. The optical frequency is $\omega=2\pi c/\lambda$, with $c$ the speed of light and $\lambda$ the wavelength of the incident radiation.

In equation (1), the polarizability is corrected by the dynamic depolarization (second addend in the denominator) and the radiative damping (third added in the denominator), with $k=2 \pi \epsilon_\textrm{m}/\lambda$ being the wave vertor.

The dielectric constant of the MNP is calculated via Drude model, with a correction applied to Au MNPs as in Derkachova, A.; Kolwas, K.; Demchenko, I. Plasmonics 2016, 11, 941-951 ([DOI:10.1007/s11468-015-0128-7](https://link.springer.com/article/10.1007/s11468-015-0128-7)):

$
\epsilon_\mathrm{p}(\omega)=\epsilon_\infty-\displaystyle{\frac{\omega_\textrm{p}^2}{\omega^2+i\omega(\gamma_\textrm{bulk}+Cv_\textrm{F}/a)}+i\Delta\epsilon(\omega)}\,\,\,\,\,\,\,\,\,\,(3)
$

with $\epsilon_\infty$ the high frequency limit, $\omega_\textrm{p}$ the plasma frequency, $\gamma_\textrm{bulk}$ the bulk dumping term,$v_\textrm{F}$ the Fermi velocity, $C$ an adimensional parameter of O(1). Finally, the correction

$
\Delta\epsilon(\omega)=\displaystyle{\frac{A}{1+e^{-(\omega-\omega_C)/\Delta}}}\,\,\,\,\,\,\,\,\,\,(4)
$

account for interbands ransitions, with $\omega_c$ and $\Delta$ are empirical parmeters to regulate the shape of the correction.

## **1.C Quantum yields**
The quantum yelds referred to the quantum efficienty without the presence of the MNP ($Q_0$) are given by

$
\displaystyle{\frac{Q}{Q_0}=}\frac{\Gamma^\textrm{(rad)}_X}{Q_0\Gamma^\textrm{(tot)}_{X}+(1-Q_0)}\,\,\,\,\,\,\,\,\,\,(5)
$

where $X=\perp,\parallel,\textrm{avg}$ respectively for the perpendicular (tangential), parallel (radial), or averaged orientation of the molecular dipole moment.

Following Vielma, J.; Leung, P. T. J. Chem. Phys. 2007, 126, 194704 ([DOI:10.1063/1.2734549](https://pubmed.ncbi.nlm.nih.gov/17523825/)) the rates can be easily calculated as follows.

### ***Tangential (parallel) molecular dipole***
The radiative contribution is

$
\displaystyle{
  \frac{\Gamma_\parallel^\textrm{(rad)}}{\Gamma_0}=\left|1-\frac{\alpha}{r^3}\right|^2
}\,\,\,\,\,\,\,\,\,\,(6)
$

and the total contribution is

$
\displaystyle{
  \frac{\Gamma_\parallel^\textrm{(tot)}}{\Gamma_0}=1+\frac{3}{4}\frac{1}{k^3}\sum_{l=1}^\infty l(l+1)\frac{\mathcal{I}m(\alpha_l)}{r^{2l+4}}
}\,\,\,\,\,\,\,\,\,\,(7)
$

with

$
\alpha_l=\displaystyle{
  \frac{\epsilon_\textrm{p}'-1}{\epsilon_\textrm{p}'+\frac{l+1}{l}}a^{2l+1}
}
$

### ***Radial (perpendicular) molecular dipole***
The radiative contribution is

$
\displaystyle{
  \frac{\Gamma_\perp^\textrm{(rad)}}{\Gamma_0}=\left|1+2\frac{\alpha}{r^3}\right|^2
}\,\,\,\,\,\,\,\,\,\,(8)
$

and the total contribution is

$
\displaystyle{
  \frac{\Gamma_\parallel^\textrm{(tot)}}{\Gamma_0}=1+\frac{3}{2}\frac{1}{k^3}\sum_{l=1}^\infty (l+1)^2\frac{\mathcal{I}m(\alpha_l)}{r^{2l+4}}
}\,\,\,\,\,\,\,\,\,\,(9)
$

### ***Averaging***
The averaged rates are

$
\Gamma^\textrm{(rad)}_\textrm{avg}=\displaystyle{\frac{1}{3}}\Gamma^\textrm{(rad)}_\perp+\frac{2}{3}\Gamma^\textrm{(rad)}_\parallel
$

and

$
\Gamma^\textrm{(tot)}_\textrm{avg}=\displaystyle{\frac{1}{3}}\Gamma^\textrm{(tot)}_\perp+\frac{2}{3}\Gamma^\textrm{(tot)}_\parallel
$

# **4 Implementation of basic equations**

```python
# @title Libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.constants import c, pi
from scipy.special import spherical_jn, spherical_yn
```

```python
# @title Utilities

# Metal NP permettivity (https://link.springer.com/article/10.1007/s11468-015-0128-7)
def NPeps(a, wavelength):
    hbar = 1.055e-34
    eV = 1.602e-19
    omega_p = widget_row.children[1].value * eV / hbar  # rad/s
    gamma_bulk = widget_row.children[2].value * eV / hbar
    vF = widget_row.children[3].value  # m/s
    C = widget_row.children[4].value

    omega = 2 * pi * c / (wavelength)
    gamma = gamma_bulk + C * vF / a  # damping with surface scattering
    eps_inf = widget_row.children[0].value # typical value for Au

    eps = eps_inf - (omega_p ** 2) / (omega ** 2 + 1j * omega * gamma)

    A = widget_row.children[5].value
    wc = widget_row.children[6].value * eV / hbar
    delta = widget_row.children[7].value * eV / hbar
    deltaEps = A / (1.0 + np.exp(-(omega - wc) / delta))

    return (eps + 1j * deltaEps)

# GN Model

def GN_decay_rates(r_em_meters, a_meters, eps_p, eps_m, wavelength_meters, l_max):

    wavelength = wavelength_meters * 1.0e9 # nm
    a = a_meters * 1.0e9 # nm
    eps = eps_p / eps_m
    k = 2 * pi  * np.sqrt(eps_m) / wavelength
    r_em = r_em_meters * 1.0e9 # nm

    Q_perp = []
    Q_para = []
    Q_avg =  []

    gamma_rad_perp_list = []
    gamma_rad_para_list = []
    gamma_rad_avg_list  = []
    gamma_tot_perp_list = []
    gamma_tot_para_list = []
    gamma_tot_avg_list  = []

    alpha1= a**3 * (eps - 1) / (eps + 2)
    alpha1 /= (1 - 1j * (2/3) * k**2 * a**2 - 1j * k**3 * alpha1 / (6*pi)) # radiative damping and dynamic depolarization


    for r in r_em:
        kr = k * r

        gamma_rad_para = np.abs(1 - alpha1 / r**3)**2
        gamma_rad_perp = np.abs(1 + 2 * alpha1 / r**3)**2

        sum_para = 0
        sum_perp = 0

        for l in range(1, l_max + 1):

            alphal = a**(2 * l + 1) * (eps - 1) / (eps + (l + 1)/l)

            sum_para += l * (l + 1) * np.imag(alphal) / r**(2 * l + 4)
            sum_perp += (l + 1)**2 * np.imag(alphal) / r**(2 * l + 4)

        gamma_tot_para = 1 + (0.75 / k**3) * sum_para
        gamma_tot_perp = 1 + (1.50 / k**3) * sum_perp

        gamma_rad_avg = (1./3.) * gamma_rad_perp + (2./3.) * gamma_rad_para
        gamma_tot_avg = (1./3.) * gamma_tot_perp + (2./3.) * gamma_tot_para

        gamma_rad_perp_list.append(gamma_rad_perp)
        gamma_rad_para_list.append(gamma_rad_para)
        gamma_rad_avg_list.append(gamma_rad_avg)

        gamma_tot_perp_list.append(gamma_tot_perp)
        gamma_tot_para_list.append(gamma_tot_para)
        gamma_tot_avg_list.append(gamma_tot_avg)

        # Quantum yilds

        Q0 = qe_w.value

        Q_perp.append(gamma_rad_perp / (Q0 * gamma_tot_perp + (1 - Q0)))
        Q_para.append(gamma_rad_para / (Q0 * gamma_tot_para + (1 - Q0)))
        Q_avg.append(gamma_rad_avg / (Q0 * gamma_tot_avg + (1 - Q0)))

    return np.array(Q_perp), np.array(Q_para), np.array(Q_avg), np.array(gamma_rad_perp_list), np.array(gamma_tot_perp_list), np.array(gamma_rad_para_list), np.array(gamma_tot_para_list), np.array(gamma_rad_avg_list), np.array(gamma_tot_avg_list)
```

# **5 Run and plot**

```python
# @title Run

global r_surf, Q_perp, Q_para, Q_avg, G_r_perp, G_t_perp, G_r_para, G_t_para, G_r_avg, G_t_avg

def run():

  # Radiation
  wavelength = lambda_w.value * 1.0e-9;
  k = 2 * pi / wavelength
  Q0 = qe_w.value

  # NP geometry
  a = 0.5 * np_diameter_w.value * 1.0e-9;
  r_surf = np.linspace(em_dist_w.value[0] * 1.0e-9, em_dist_w.value[1] * 1.0e-9, 200)
  r_em = r_surf + a # emitter distance range

  # Medium
  n_med = ref_index_medium_w.value
  eps_m = n_med**2

  # NP permittivity
  eps_p = NPeps(a, wavelength)

  # Lmax
  l_max = L_w.value

  print("NP epsilon = ", eps_p)

  # Lmax
  l_max = L_w.value

  Q_perp, Q_para, Q_avg, G_r_perp, G_t_perp, G_r_para, G_t_para, G_r_avg, G_t_avg = GN_decay_rates(r_em, a, eps_p, eps_m, wavelength, l_max)

  return r_surf, Q_perp, Q_para, Q_avg, G_r_perp, G_t_perp, G_r_para, G_t_para, G_r_avg, G_t_avg

# Create a button to run the calculation
run_button = widgets.Button(description="RUN CALCULATION")
run_output = widgets.Output()

display(run_button, run_output)

from IPython.display import clear_output
def on_run_button_clicked(b):
    global r_surf, Q_perp, Q_para, Q_avg, G_r_perp, G_t_perp, G_r_para, G_t_para, G_r_avg, G_t_avg  # Declare global variables here as well
    with run_output:
        run_output.clear_output(wait=True)
        print("Running calculation...")
        # Assign the returned values to the global variables
        r_surf, Q_perp, Q_para, Q_avg, G_r_perp, G_t_perp, G_r_para, G_t_para, G_r_avg, G_t_avg = run()
        print("Calculation complete")


run_button.on_click(on_run_button_clicked)
```

```python
# @title Plot results
def plot_results(r_surf, Q_perp, Q_para, Q_avg):
  plt.figure(figsize=(8,5))
  plt.plot(r_surf*1e9, 100*Q_perp, label='Perpendicular component')
  plt.plot(r_surf*1e9, 100*Q_para, label='Parallel component')
  plt.plot(r_surf*1e9, 100*Q_avg, label='Average orientation')
  plt.xlabel('Distance from sphere surface (nm)')
  plt.ylabel('Quantum yield %')
  plt.title('Quantum yield vs distance')
  plt.grid(True)
  plt.legend()
  plt.tight_layout()
  plt.show()

# Create a button to plot the results
plot_button = widgets.Button(description="PLOT RESULTS")
plot_output = widgets.Output()

display(plot_button, plot_output)

from IPython.display import clear_output
def on_plot_button_clicked(b):
    with plot_output:
        plot_output.clear_output(wait=True)
        # Check if r_surf is defined before plotting
        if 'r_surf' in globals() and r_surf is not None:
            plot_results(r_surf, Q_perp, Q_para, Q_avg)
        else:
            print("Please run the calculation first.")


plot_button.on_click(on_plot_button_clicked)
```

```python
# @title Save results CSV

# prompt: Can you modify the previous cell (sare results) to save data in CSV format (comma separated), including also gamma_rad and gamma_tot? The saved file should auomatically download

import csv
from google.colab import files

def save_results_csv(r_surf, Q_perp, Q_par, Q_avg, gamma_rad_perp, gamma_rad_para, gamma_tot_perp, gamma_tot_para, gamma_rad_avg, gamma_tot_avg, filename):
    """
    Saves the calculation results to a CSV file including gamma_rad and gamma_tot,
    and automatically downloads the file.

    Args:
        r_surf (np.ndarray): Distance from the sphere surface in nm.
        Q_perp (np.ndarray): Perpendicular quantum yield.
        Q_par (np.ndarray): Parallel quantum yield.
        Q_avg (np.ndarray): Average quantum yield.
        gamma_rad_perp (np.ndarray): Perpendicular radiative rate.
        gamma_rad_para (np.ndarray): Parallel radiative rate.
        gamma_tot_perp (np.ndarray): Perpendicular total rate.
        gamma_tot_para (np.ndarray): Parallel total rate.
        gamma_rad_avg (np.ndarray): Average radiative rate.
        gamma_tot_avg (np.ndarray): Average total rate.
        filename (str): The name of the file to save the results to.
    """
    # Ensure r_surf is in nm for saving
    r_surf_nm = r_surf * 1e9

    data = zip(r_surf_nm, Q_perp, Q_par, Q_avg, gamma_rad_perp, gamma_rad_para, gamma_tot_perp, gamma_tot_para, gamma_rad_avg, gamma_tot_avg)

    header = ["r_surf (nm)", "Q_perp", "Q_para", "Q_avg", "gamma_rad_perp", "gamma_rad_para", "gamma_tot_perp", "gamma_tot_para", "gamma_rad_avg", "gamma_tot_avg"]

    with open(filename, 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(header)
        writer.writerows(data)

    print(f"Results saved to {filename}")
    files.download(filename)


# Create a button to save the results
fname_w = widgets.Text(
    value='yields.csv',
    placeholder='File name:',
    description='File name:',
    disabled=False
)
display(fname_w)


save_button_csv = widgets.Button(description="SAVE CSV")
save_output_csv = widgets.Output()

display(save_button_csv, save_output_csv)

from IPython.display import clear_output
def on_save_button_csv_clicked(b):
    print("Save button clicked - Entering on_save_button_csv_clicked") # Added for debugging
    with save_output_csv:
        save_output_csv.clear_output(wait=True)
        print("File saving...")
        save_results_csv(r_surf, Q_perp, Q_para, Q_avg, G_r_perp, G_r_para, G_t_perp, G_t_para, G_r_avg, G_t_avg, fname_w.value)

save_button_csv.on_click(on_save_button_csv_clicked)
```
